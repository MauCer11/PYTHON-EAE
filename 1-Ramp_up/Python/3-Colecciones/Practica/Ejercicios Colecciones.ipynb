## Exercise 4
Given the following list, find the index of value 45 and replace it with 0
> ```ejer_4 = [20, 47, 19, 29, 45, 67, 78, 90]```

````python
ejer_4 = [20, 47, 19, 29, 45, 67, 78, 90]

for num in ejer_4:
    if num == 45:
        index=ejer_4.index(num)
        ejer_4[index] = 0
        print(f"number 45 found at index {index}")        
print(ejer_4)
````

## Exercise 5
Given the following list, delete all values equal to 3
> ```ejer_5 = [3, 20, 3, 47, 19, 3, 29, 45, 67, 78, 90, 3, 3]```

TIP: Do not try to delete elements from the list you are iterating. Make a copy with `ejer_5.copy()`.

````python
##Este codigo no es correcto porque al eliminar un elemento, el índice de los elementos siguientes cambia, lo que puede llevar a que se salten algunos elementos o se eliminen incorrectamente. En este caso, al eliminar el número 3, el siguiente número 3 se moverá a la posición del número eliminado, lo que hará que el bucle lo omita y no se elimine.
ejer_5 = [3, 20, 3, 47, 19, 3, 29, 45, 67, 78, 90, 3, 3]
for num in ejer_5:
    if num == 3:
        ejer_5.remove(num)
print(ejer_5)




````

````python
## Una forma correcta es usar una copia de la lista para iterar sobre ella mientras se modifica la lista original:

ejer_5 = [3, 20, 3, 47, 19, 3, 29, 45, 67, 78, 90, 3, 3]
for num in ejer_5.copy():
    if num == 3:
        ejer_5.remove(num)

print(ejer_5)
````

````python
## Otra forma es usar while para eliminar todos los elementos 3 de la lista:
ejer_test = [3, 20, 3, 47, 19, 3, 29, 45, 67, 78, 90, 3, 3]

while 3 in ejer_test:
    ejer_test.remove(3)
print(ejer_test)
````

## Exercise 14
Merge both dictionaries. To do this, use `update`
> ```ejer_14_1 = {1: 11, 2: 22}```
>
> ```ejer_14_2 = {3: 33, 4: 44}```

````python
ejer_14_1 = {1: 11, 2: 22}
ejer_14_2 = {3: 33, 4: 44}

ejer_14_merged = {**ejer_14_1, **ejer_14_2}
print(ejer_14_merged)

ejer_14_1.update(ejer_14_2)
print(ejer_14_1)
````

## Exercise 15
Sum all values of the dictionary
> ```ejer_15 = {1: 11, 2: 22, 3: 33, 4: 44, 5: 55}```

````python
ejer_15 = {1: 11, 2: 22, 3: 33, 4: 44, 5: 55}
sum_values=0
for values in ejer_15.values():
    sum_values += values
print(sum_values)
````
