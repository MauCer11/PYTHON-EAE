## Exercise 1
We are a company that has several appliance stores. We need a program to manage the stores, see the sales they have had, how many employees there are, etc... For this we are going to model the store in a class, which will have the following characteristics:
* Class name: "Store"
* Common attributes:
    * Type: "Appliances"
    * Open: True
* Attributes specific to each store:
    * Name - String
    * Address - String
    * Number of employees - int
    * Sales last 3 months - List of 3 numbers
* Method to calculate sales from all months, returning a number with all sales.
* Method that calculates the average sales of the last months, per employee
* Method that returns in a string the name of the store, along with its address.
* Method that gets the sales from the last month.
* Method to give a sales projection in case X money has been invested in marketing for that store. Where X is the input parameter. If X is less than 1000, sales for the last 3 months would have been (1.2 x sales), if greater than or equal to 1000, sales would have been (1.5 x sales). The method rewrites the sales attribute for the last 3 months, and also returns the sales with the increase.

It is requested:
1. Implement the "Store" class
2. Create three stores with invented data
3. Check in at least one of them, everything implemented in the store class (its attributes, average sales, sales/employee...)
3. Calculate the sales from the last month of all stores. For this use the `for` loop
4. Print on screen the names of the stores whose address contains the string "Avenue"
5. I'm sure you've already done this section... Document the class :)

```python

class store():
    Type = "Appliances"
    open = True

    def __init__(self,name,address,employees_num,last_3_months_sales):
        self.name = name
        self.address = address
        self.employees_num = employees_num
        self.last_3_months = last_3_months_sales
    
    def total_3_months(self):
       return sum(self.last_3_months)
    def average_per_employees(self):
        return sum(self.last_3_months) / self.employees_num
    def name_address(self):
        return (f"store: {self.name} is located in {self.address} ")
    def last_month(self):
        return self.last_3_months[-1]
    def sales_projection(self):
        investment = int(input("X = "))
        total_proj=0
        if investment < 1000:
            total_proj = sum(self.last_3_months) * 1.2
        if investment >= 1000:
            total_proj = sum(self.last_3_months) * 1.5
        return total_proj


  


store_1 = store("amazon","Principe de vergara 205", 500, [10000,25000,32002])
store_2 = store("zara", "Avenue joaquin costa 532", 58, [589,874,965])
store_3 = store("pull & bear", "sainz de baranda avenue 56", 70, [5890,8740,9650])

print(store_1.sales_projection())
print("--------------------------")
print(store_1.total_3_months())
print(store_1.average_per_employees())
print(store_1.name_address())

print(store_1.name_address())
print(store_2.average_per_employees())
print(store_1.last_month())
print(store_1.name)
print("----------------------------------------")

# 3. Calculate the sales from the last month of all stores. For this use the `for` loop
all_stores = [store_1,store_2,store_3]
for i in all_stores:
    print(i.last_month())
print("----------------------------------------")

# 5. Print on screen the names of the stores whose address contains the string "Avenue"
for i in all_stores:
    if "avenue" in i.address.lower():
        print(i.name)

  
```

## Exercise 2
In this exercise we are going to implement a *Dog* class in Python. The class has the following characteristics:
* Things we know for sure a dog has
    * It has 4 legs
    * 2 ears
    * 2 eyes
    * A speed of 0. By default, the dog is stopped
* When initialized:
    * The dog will be of a certain breed
    * By default it will have "Brown" fur, unless otherwise stated.
    * By default it will have no owner, unless otherwise stated.
    
* It will also have a method called walk, which has an input argument (speed_increase). This value will be added to the speed the dog already had.
* It needs another method (stop), where we will set the speed to 0.
* Another method called "bark", which will have an input argument, and the output will be the following string: "WOOF!" + the input argument.


It is requested:
* Implement the *Dog* class
* Create an object of type *Dog*, without an owner
* Check that all its attributes are correct
* Test that it walks, and check its speed
* Stop it
* Document the *Dog* class

```python
class dog():
    """The common characteristics of a dog are: legs, ears, eyes and speed."""
    legs = 4
    ears = 2
    eyes = 2
    speed = 0

    def __init__(self,breed,fur = "brown", owner = None):
        self.breed = breed
        self.fur = fur
        self.owner = owner
    
    def characteristics(self):
        return (f"breed = {self.breed} \n fur = {self.fur} \n owner = {self.owner}" )
    def walk(self):
        speed_increase = int(input("speed = "))
        self.speed = speed_increase + self.speed
        return self.speed
    def stop(self):
        self.speed = 0
        return self.speed
    def bark(self,message):
        #barking = input("is the dog barking?").lower()
        return (f"WOOF! {message}!!")



dog_1= dog("pastor aleman")
dog_2= dog("gran danes","white")
dog_3= dog("chuhuahua",owner ="Mau")

print(dog_3.characteristics())
dog_3.walk()
print(dog_3.speed)
dog_3.walk()
print(dog_3.speed)
dog_3.stop()
print(dog_3.speed)
dog_3.bark("Hola")

print(dog.__doc__)
#print(dog.__init__.__doc__)
#print(help(dog))



```
