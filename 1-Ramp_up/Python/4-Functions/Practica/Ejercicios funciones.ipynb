## Exercise 1
Write a function that converts numbers from 1 to 7 into names of the days of the week. The function will consist of a single numeric argument and a string type output

````python
def week_days(*days):
    for val in days:
        if val == 1:
            print("Monday")
        elif val == 2:
            print("Tuesday")
        elif val == 3:
            print("Wednesday")
        elif val == 4:
            print("Thursday")
        elif val == 5:
            print("Friday")
        elif val == 6:
            print("Saturday")
        elif val == 7:
            print("Sunday")
        else:
            print("Invalid day")
week_days(6, 7, 3, 3, 5, 9)
````

## Exercise 3
Write a function that compares two numbers. The function has two arguments and there are three possible outputs: they are equal, the first is greater than the second, or the second is greater than the first

````python
def compare(num1, num2):
    if num1 == num2:
        print("they are equal")
    elif num1 > num2:
        print(f"number {num1} is greater than number {num2}")
    elif num2 > num1:
        print(f"number {num1} is smaller than number {num2}")

compare(88,88)
````

## Exercise 5
Write a function that has a single argument, a string. The output of the function must be a dictionary with the count of all the letters of that string.

````python
def count_letters(string):
    dic_1={}
    for index, letter in enumerate(string):
        dic_1[index +1]= letter
    print(dic_1)

count_letters("Hola como estas?")

````

## Exercise 6
Write a function that adds or removes elements in a list. The function needs the following arguments:
* list: the list where elements will be added or removed
* command: "add" or "remove"
* element: By default it is None.

````python
def add_remove(list_1,command_1,element= None):
    list_function=[]
    for elem in list_1:
            list_function.append(elem)
    if command_1 == "add":
        
        list_function.append(element)
        
    
    elif command_1 == "remove":
        if element in list_function:
             list_function.remove(element)
        else:
             print(f"No element '{element}' to remove in the list")

    return list_function

lista_M=["queso","jamon","tortillas"]
add_remove(lista_M,"remove", "tortillas")
    
    
````

## Exercise 7
Create a function that receives an arbitrary number of words, and returns a complete sentence, separating the words with spaces.

````python
def sentence(*words):
    sentence = ""
    for word in words:
        sentence = sentence + " "+ word
        
    return sentence [1:]

sentence("hola","como","estas")
````

## Exercise 8
Write a program that obtains the nth number of the [Fibonacci sequence](https://en.wikipedia.org/wiki/Fibonacci_number). You must create a recursive function with a single argument.

```Python
fibonacci(2) = 1
fibonacci(4) = 3
fibonacci(5) = 5
fibonacci(8) = 21
```

````python
def fibonacci(num):
    i=0
    j=0
    fib = [0,1]
    sum_1 = 0
    if num < 0:
        print("type a positive number")
   
    while i<num:
        sum_1 = fib[j]+fib[j+1]
        fib.append(sum_1)
        i += 1
        j += 1
        
    
    return fib[num]

fibonacci(1)
````

* Two circles with radius 10 + a triangle with base 3 and height 7
* A square with side = 10 + 3 circles (one with radius = 4 and the other two with radius = 6) + 5 triangles with base = 2 + height = 4

````python
import math
def square_area(side):
    sq_area = side**2
    return sq_area
def triangle_area(base,height):
    tri_area = base * height / 2
    return tri_area
def circle_area(radius):
    circ_area = math.pi * radius**2
    return circ_area
````
