## Dataset Information

The Titanic dataset contains information about passengers on the Titanic.

**Columns:**
- `pclass`: Passenger class (1 = First, 2 = Second, 3 = Third)
- `survived`: Survival (0 = No, 1 = Yes)
- `name`: Passenger name
- `sex`: Gender (male/female)
- `age`: Age in years
- `sibsp`: Number of siblings/spouses aboard
- `parch`: Number of parents/children aboard
- `ticket`: Ticket number
- `fare`: Passenger fare
- `cabin`: Cabin number
- `embarked`: Port of embarkation (C = Cherbourg, Q = Queenstown, S = Southampton)
- `boat`: Lifeboat number
- `body`: Body identification number
- `home.dest`: Home/destination

## Question 5: Multiple Conditions (1.5 points)

**Tasks:**

a) Display all passengers who were in first class (pclass = 1) AND survived (survived = 1). Show the complete DataFrame. (0.7 points)

b) How many passengers were in first class AND survived? Display just the number. (0.4 points)

c) Display all passengers who were either children (age < 12) OR elderly (age > 60). Show the complete DataFrame. (0.4 points)

**Important:** Remember to use `&` for AND and `|` for OR, with parentheses around each condition!

````python
# a) First class survivors
titanic[(titanic["pclass"] == 1) & (titanic["survived"] == 1)]
````

````python
# b) Count of first class survivors
titanic[(titanic["pclass"] == 1) & (titanic["survived"] == 1)].shape[0]
````

````python
# c) Children or elderly passengers
titanic[(titanic["age"] < 12) | (titanic["age"] > 60)]
````

## Question 6: Using .isin() Method (1 point)

**Tasks:**

a) Display all passengers who embarked from either Cherbourg ('C') or Queenstown ('Q'). Show the complete DataFrame. (0.6 points)

b) Count how many passengers embarked from Cherbourg or Queenstown. Display just the number. (0.4 points)

**Hint:** Use the `.isin()` method with a list of values

## Question 7: Column Selection with .loc[] (1 point)

**Tasks:**

a) Display only the columns `name`, `age`, and `survived` for all male passengers who survived. (0.6 points)

b) Display only the `name` and `fare` columns for passengers in third class (pclass = 3). (0.4 points)

**Important:** Use `.loc[]` for these selections

````python
# a) Name, age, survived for male survivors
titanic.loc[(titanic["survived"]==1) & (titanic["sex"]=="male"), ["name", "age", "survived"]]
````

````python
# b) Name and fare for third class
titanic.loc[titanic["pclass"]==3, ["name", "fare"]]
````

## Question 8: Finding Maximum and Minimum Values (1.5 points)

**Tasks:**

a) What was the maximum fare paid? Display just the number. (0.3 points)

b) Display the complete row for the passenger who paid the maximum fare. Show all columns for this passenger. (0.5 points)

c) What was the minimum age of passengers (excluding missing values)? Display just the number. (0.3 points)

d) Display the complete row for the youngest passenger. Show all columns. (0.4 points)

**Hint for b and d:** Use `.idxmax()` or filter where value equals the max/min

````python
# a) Maximum fare
titanic["fare"].max()
````

````python
# b) Passenger who paid maximum fare with max
max_fare = titanic["fare"].max()
titanic[titanic["fare"] == max_fare]
````

````python
# b) Passenger who paid maximum fare with idxmax
titanic.loc[titanic["fare"].idxmax()]
````

````python
# c) Minimum age
titanic["age"].min()
````

````python
# d) Youngest passenger with idxmin
titanic.loc[titanic["age"].idxmin()]
````

````python
# d) Youngest passenger with min
min_age= titanic["age"].min()
titanic[titanic["age"] == min_age]
````

**Task:**

Calculate the **survival rate** (percentage of survivors) for each passenger class.

Your answer should show:
- The passenger class (1, 2, or 3)
