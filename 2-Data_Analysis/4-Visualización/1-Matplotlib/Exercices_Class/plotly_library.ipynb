## Part 2: Your First Plot - Line Chart

Let's create a simple line chart to understand the basics.

````python
# Create sample data
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

# Create a line chart
fig = go.Figure(data=go.Scatter(x=x, y=y, mode='lines'))

# Display the chart
fig.show()
````

**What happened here?**

1. **`go.Figure()`**: Creates a new figure (empty canvas)
2. **`go.Scatter()`**: Creates a scatter/line trace
3. **`x=x, y=y`**: Provides the data points
4. **`mode='lines'`**: Draws lines between points
5. **`fig.show()`**: Displays the interactive chart

**Try this:** Hover over the line to see values!

### Adding Markers and Customization

````python
# Line with markers
fig = go.Figure(data=go.Scatter(
    x=x, 
    y=y, 
    mode='lines+markers',  # Both lines AND markers
    name='My First Line',   # Legend name
    line=dict(color='purple', width=2),  # Line style
    marker=dict(size=10, color='red')  # Marker style
))

fig.show()
````

**New concepts:**

- **`mode='lines+markers'`**: Shows both lines and points
- **`name`**: Appears in the legend
- **`line=dict(...)`**: Customize line color and width
- **`marker=dict(...)`**: Customize marker size and color

---
## Part 3: Adding Titles and Labels

````python
# Create a chart with proper labels
fig = go.Figure(data=go.Scatter(x=x, y=y, mode='lines+markers'))

# Update layout with titles and labels
fig.update_layout(
    title='My First Plotly Chart',
    xaxis_title='X Axis Label',
    yaxis_title='Y Axis Label',
    font=dict(size=14)
)

fig.show()
````

**`.update_layout()` is powerful!**

It controls:
- Titles and labels
- Colors and themes
- Size and margins
- Legend position
- Much more!

## Part 4: Bar Charts

````python
# Sample data: fruit sales
fruits = ['Apples', 'Oranges', 'Bananas', 'Grapes', 'Pears']
sales = [45, 32, 28, 50, 35]

# Create bar chart
fig = go.Figure(data=go.Bar(x=fruits, y=sales))

fig.update_layout(
    title='Fruit Sales',
    xaxis_title='Fruit',
    yaxis_title='Units Sold'
)

fig.show()
````

**Key difference:**
- Use **`go.Bar()`** instead of `go.Scatter()`
- Everything else works the same way!

### Customizing Bar Colors

````python
# Bar chart with custom colors
fig = go.Figure(data=go.Bar(
    x=fruits, 
    y=sales,
    marker=dict(
        color=['red', 'orange', 'yellow', 'purple', 'green'],  # One color per bar
        line=dict(color='black', width=2)  # Bar border
    )
))

fig.update_layout(title='Colorful Fruit Sales')
fig.show()
````

---
## Part 5: Multiple Traces (Lines/Bars)

You can add multiple datasets to the same chart!

````python
# Data for two products
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
product_a = [20, 25, 30, 28, 35]
product_b = [15, 22, 27, 30, 33]

# Create figure
fig = go.Figure()

# Add first trace
fig.add_trace(go.Scatter(
    x=months, 
    y=product_a, 
    mode='lines+markers',
    name='Product A',
    line=dict(color='blue')
))

# Add second trace
fig.add_trace(go.Scatter(
    x=months, 
    y=product_b, 
    mode='lines+markers',
    name='Product B',
    line=dict(color='red')
))

fig.update_layout(
    title='Product Sales Comparison',
    xaxis_title='Month',
    yaxis_title='Sales'
)

fig.show()
````

**New method: `.add_trace()`**

- Create an empty figure: `fig = go.Figure()`
- Add traces one by one: `fig.add_trace(...)`
- Each trace appears in the legend

**Try this:** Click on legend items to hide/show traces!

## Part 6: Scatter Plots

````python
# Generate random data
np.random.seed(42)
x_random = np.random.randn(100)
y_random = np.random.randn(100)

# Create scatter plot
fig = go.Figure(data=go.Scatter(
    x=x_random,
    y=y_random,
    mode='markers',  # Only markers, no lines
    marker=dict(
        size=10,
        color=y_random,  # Color by y values
        colorscale='Viridis',  # Color scheme
        showscale=True  # Show color bar
    )
))

fig.update_layout(
    title='Scatter Plot with Color Scale',
    xaxis_title='X values',
    yaxis_title='Y values'
)

fig.show()
````

**Color scales:**
- **`color=y_random`**: Uses data to determine colors
- **`colorscale='Viridis'`**: Color scheme name
- **`showscale=True`**: Displays the color bar

**Other color scales:** 'Plasma', 'Jet', 'Rainbow', 'Blues', 'Reds'

## Part 7: Plotly Express - The Easy Way!

Plotly Express (`px`) makes common charts with just one line!

````python
# Create sample DataFrame
df = pd.DataFrame({
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    'Sales': [150, 180, 220, 210, 250, 280],
    'Expenses': [100, 110, 130, 140, 150, 160]
})

print(df)
````

````python
# Create line chart with Plotly Express
fig = px.line(df, x='Month', y='Sales', title='Sales Over Time')
fig.show()
````

**That's it! Just one line!**

- **`px.line()`**: Creates a line chart
- **`df`**: Your DataFrame
- **`x='Month'`**: Column for x-axis
- **`y='Sales'`**: Column for y-axis

### Multiple Lines with Plotly Express

````python
# Reshape data for multiple lines
df_long = df.melt(id_vars='Month', value_vars=['Sales', 'Expenses'],
                  var_name='Category', value_name='Amount')

print(df_long.head())
````

````python
# Plot multiple lines automatically
fig = px.line(df_long, x='Month', y='Amount', color='Category',
              title='Sales vs Expenses')
fig.show()
````

**Magic parameter: `color='Category'`**

- Automatically creates one line per category
- Different colors for each
- Legend is automatic!

---
## Part 8: Bar Charts with Plotly Express

````python
# Simple bar chart
fig = px.bar(df, x='Month', y='Sales', title='Monthly Sales')
fig.show()
````

### Grouped Bar Chart

````python
# Grouped bars
fig = px.bar(df_long, x='Month', y='Amount', color='Category',
             barmode='group',  # Group bars side by side
             title='Sales vs Expenses (Grouped)')
fig.show()
````

### Stacked Bar Chart

````python
# Stacked bars
fig = px.bar(df_long, x='Month', y='Amount', color='Category',
             barmode='stack',  # Stack bars on top
             title='Sales vs Expenses (Stacked)')
fig.show()
````

**`barmode` options:**
- **`'group'`**: Side by side
- **`'stack'`**: On top of each other
- **`'relative'`**: For positive/negative values

## Part 9: Pie Charts

````python
# Market share data
companies = ['Company A', 'Company B', 'Company C', 'Company D']
market_share = [35, 25, 20, 20]

# Create pie chart
fig = go.Figure(data=go.Pie(
    labels=companies,
    values=market_share
))

fig.update_layout(title='Market Share')
fig.show()
````

**Try this:** Click and drag a slice to pull it out!

````python
# Pie chart with Plotly Express
df_pie = pd.DataFrame({
    'Company': companies,
    'Share': market_share
})

fig = px.pie(df_pie, names='Company', values='Share', 
             title='Market Share',
             hole=0.3)  # Creates a donut chart!
fig.show()
````

**`hole=0.3`** creates a donut chart (30% hole in center)

## Part 10: Histograms

````python
# Generate random normal distribution
data = np.random.randn(1000)

# Create histogram with Plotly Express
fig = px.histogram(data, nbins=30, title='Distribution of Random Data')
fig.show()
````

**`nbins`** controls the number of bins (bars)

````python
# Histogram with custom styling
fig = go.Figure(data=go.Histogram(
    x=data,
    nbinsx=30,
    marker=dict(color='lightblue', line=dict(color='black', width=1))
))

fig.update_layout(
    title='Custom Histogram',
    xaxis_title='Value',
    yaxis_title='Frequency'
)

fig.show()
````

## Part 11: Box Plots

````python
# Create sample data for different groups
np.random.seed(42)
group_a = np.random.normal(100, 15, 200)
group_b = np.random.normal(110, 20, 200)
group_c = np.random.normal(90, 10, 200)

# Create DataFrame
df_box = pd.DataFrame({
    'Group A': group_a,
    'Group B': group_b,
    'Group C': group_c
})

# Melt for Plotly Express
df_box_long = df_box.melt(var_name='Group', value_name='Value')

# Create box plot
fig = px.box(df_box_long, x='Group', y='Value', 
             title='Distribution Comparison')
fig.show()
````

**Box plots show:**
- Median (line in the box)
- Quartiles (box edges)
- Outliers (individual points)
- Range (whiskers)

## Part 12: Subplots - Multiple Charts in One Figure

````python
from plotly.subplots import make_subplots

# Create subplots: 1 row, 2 columns
fig = make_subplots(rows=1, cols=2, 
                    subplot_titles=('Chart 1', 'Chart 2'))

# Add first chart (left)
fig.add_trace(
    go.Scatter(x=[1, 2, 3], y=[4, 5, 6], name='Line 1'),
    row=1, col=1
)

# Add second chart (right)
fig.add_trace(
    go.Bar(x=['A', 'B', 'C'], y=[10, 20, 15], name='Bars'),
    row=1, col=2
)

fig.update_layout(title='Subplots Example', showlegend=True)
fig.show()
````

**Steps for subplots:**
1. Import: `from plotly.subplots import make_subplots`
2. Create grid: `make_subplots(rows=..., cols=...)`
3. Add traces: Specify `row=` and `col=`

````python
# 2x2 grid of different chart types
fig = make_subplots(
    rows=2, cols=2,
    subplot_titles=('Line', 'Bar', 'Scatter', 'Pie'),
    specs=[[{'type': 'scatter'}, {'type': 'bar'}],
           [{'type': 'scatter'}, {'type': 'pie'}]]
)

# Line chart (top-left)
fig.add_trace(go.Scatter(x=[1,2,3], y=[2,4,6], mode='lines'), row=1, col=1)

# Bar chart (top-right)
fig.add_trace(go.Bar(x=['A','B','C'], y=[10,15,12]), row=1, col=2)

# Scatter (bottom-left)
fig.add_trace(go.Scatter(x=[1,2,3,4], y=[10,11,12,13], mode='markers'), row=2, col=1)

# Pie (bottom-right)
fig.add_trace(go.Pie(labels=['X','Y','Z'], values=[30,40,30]), row=2, col=2)

fig.update_layout(title='Dashboard with 4 Charts', height=600)
fig.show()
````

## Part 13: Heatmaps

````python
# Create correlation matrix
df_corr = pd.DataFrame(
    np.random.randn(10, 5),
    columns=['A', 'B', 'C', 'D', 'E']
)

correlation_matrix = df_corr.corr()

# Create heatmap
fig = go.Figure(data=go.Heatmap(
    z=correlation_matrix.values,
    x=correlation_matrix.columns,
    y=correlation_matrix.columns,
    colorscale='RdBu',
    zmid=0  # Center color at 0
))

fig.update_layout(title='Correlation Heatmap')
fig.show()
````

**Heatmap parameters:**
- **`z`**: 2D array of values
- **`x, y`**: Labels for axes
- **`colorscale`**: Color scheme
- **`zmid`**: Center value for color scale

````python
# Heatmap with Plotly Express
fig = px.imshow(correlation_matrix, 
                title='Correlation Matrix',
                color_continuous_scale='Viridis',
                aspect='auto')
fig.show()
````

## Part 14: 3D Scatter Plots

````python
# Generate 3D data
np.random.seed(42)
n = 100
x3d = np.random.randn(n)
y3d = np.random.randn(n)
z3d = np.random.randn(n)
colors = np.random.randn(n)

# Create 3D scatter plot
fig = go.Figure(data=go.Scatter3d(
    x=x3d,
    y=y3d,
    z=z3d,
    mode='markers',
    marker=dict(
        size=5,
        color=colors,
        colorscale='Viridis',
        showscale=True
    )
))

fig.update_layout(
    title='3D Scatter Plot',
    scene=dict(
        xaxis_title='X Axis',
        yaxis_title='Y Axis',
        zaxis_title='Z Axis'
    )
)

fig.show()
````

## Part 15: Real-World Example - Sales Dashboard

````python
# Create realistic sales data
np.random.seed(42)
dates = pd.date_range('2024-01-01', periods=365, freq='D')
sales = 1000 + np.cumsum(np.random.randn(365) * 50)

df_sales = pd.DataFrame({
    'Date': dates,
    'Sales': sales
})

df_sales['Month'] = df_sales['Date'].dt.month
df_sales['Day_of_Week'] = df_sales['Date'].dt.dayofweek

print(df_sales.head())
````

````python
# Time series plot
fig = px.line(df_sales, x='Date', y='Sales', 
              title='Daily Sales - 2024')

fig.update_layout(
    xaxis_title='Date',
    yaxis_title='Sales ($)',
    hovermode='x unified'
)

fig.show()
````

````python
# Monthly aggregation
monthly_sales = df_sales.groupby('Month')['Sales'].mean().reset_index()
monthly_sales['Month_Name'] = pd.to_datetime(monthly_sales['Month'], format='%m').dt.strftime('%B')

fig = px.bar(monthly_sales, x='Month_Name', y='Sales',
             title='Average Sales by Month')
fig.show()
````

## Summary: Key Concepts

### Two Main Approaches

**1. Plotly Graph Objects (go)**
- More control and customization
- `go.Figure()`, `go.Scatter()`, `go.Bar()`, etc.
- Good for complex visualizations

**2. Plotly Express (px)**
- Quick and easy
- `px.line()`, `px.bar()`, `px.scatter()`, etc.
- Great for exploring data fast

### Common Chart Types

| Chart Type | Graph Objects | Plotly Express |
|------------|---------------|----------------|
| Line | `go.Scatter(mode='lines')` | `px.line()` |
| Scatter | `go.Scatter(mode='markers')` | `px.scatter()` |
| Bar | `go.Bar()` | `px.bar()` |
| Pie | `go.Pie()` | `px.pie()` |
| Histogram | `go.Histogram()` | `px.histogram()` |
| Box | `go.Box()` | `px.box()` |
| Heatmap | `go.Heatmap()` | `px.imshow()` |

### Key Methods

- **`.show()`**: Display the chart
- **`.update_layout()`**: Modify titles, colors, size
- **`.add_trace()`**: Add more data series
- **`.write_html()`**: Save as HTML

### Customization Options

```python
fig.update_layout(
    title='My Title',
    xaxis_title='X Label',
    yaxis_title='Y Label',
    font=dict(size=14, color='black'),
    plot_bgcolor='white',

```
